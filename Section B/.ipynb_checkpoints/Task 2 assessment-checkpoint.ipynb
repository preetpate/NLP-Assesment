```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

reviews = [
    "The spicy chicken was fresh and delicious",
    "Cold fries arrived late and soggy",
    "The burger was tasty and juicy",
    "Food was cold and delivery was late",
    "Fresh salad with crisp vegetables",
    "Average meal with normal portion size",
    "The pasta sauce was rich and tasty",
    "Order was correct but the packaging was damaged",
    "Very spicy noodles with fresh vegetables"
]
```

```python
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(reviews)
feature_names = vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names,
    index=[f"Review {i}" for i in range(1, len(reviews) + 1)]
)

pd.set_option("display.max_columns", None)
display(tfidf_df.round(3))
```

```python
# 2

for i, scores in enumerate(tfidf_matrix.toarray(), start=1):
    top_indices = scores.argsort()[::-1]
    top_terms = [(feature_names[j], scores[j]) for j in top_indices if scores[j] > 0][:3]
    print(f"Review {i}: " + ", ".join(f"{term} ({score:.3f})" for term, score in top_terms))
```
