```python
#2
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

STOP_WORDS = set(stopwords.words("english")) - {"not", "no", "nor"}
stemmer = PorterStemmer()

def preprocess_text(text: str) -> str:
    """Clean a review and return a stemmed, space-joined string."""
    if not isinstance(text, str):
        text = ""
    text = text.lower()
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    tokens = re.findall(r"\b\w+\b", text)
    tokens = [stemmer.stem(token) for token in tokens if token not in STOP_WORDS]
    return " ".join(tokens)
```

```python
#3
samples = [
    "The pizza arrived hot, fresh, and delicious!",
    "My burger was cold and the delivery was very late.",
    "I did not receive my drink, and the app crashed."
]

for i, review in enumerate(samples, start=1):
    print(f"Review {i}: {review}")
    print(f"Processed: {preprocess_text(review)}\\n")
```

```python
#4
custom_review = input("Enter a food delivery review: ")
print("Processed review:", preprocess_text(custom_review))
```
