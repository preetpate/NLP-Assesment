1. Exact prompt given to the AI tool

Write a simple Python program for my Data Science Module 11 NLP assessment. The program should accept a list of food delivery customer reviews, preprocess every review using tokenisation, stopword removal, and stemming, build a TF-IDF matrix, and train a Logistic Regression model to classify reviews as Positive or Negative. Print the model's accuracy using a train-test split and allow the user to enter three new reviews at runtime to predict their sentiment. Use easy-to-understand code with comments suitable for a student assignment.

```python
# 2

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

nltk.download("stopwords")

reviews = [
    "The food was delicious and delivery was fast",
    "Amazing taste and excellent service",
    "I loved the food and the delivery",
    "The meal was fresh and very tasty",
    "Great food and friendly delivery person",
    "The food was cold and tasteless",
    "Very bad service and late delivery",
    "I hated the food, it was terrible",
    "Worst meal ever, completely disappointing",
    "The delivery was slow and the food was awful",
    "Good quality food and quick service",
    "The food was not good",
    "Excellent food and fast delivery",
    "The order was damaged and disgusting",
    "Really tasty food, I enjoyed it"
]

labels = [
    "Positive", "Positive", "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative", "Negative", "Negative",
    "Positive", "Negative", "Positive", "Negative", "Positive"
]

stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

```

```python
def preprocess(text):
    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
    words = [word for word in words if word not in stop_words]
    words = [stemmer.stem(word) for word in words]
    return " ".join(words)

clean_reviews = [preprocess(review) for review in reviews]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(clean_reviews)
y = labels

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = LogisticRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, predictions))

for i in range(3):
    new_review = input("Enter a review: ")
    clean_review = preprocess(new_review)
    review_vector = vectorizer.transform([clean_review])
    sentiment = model.predict(review_vector)[0]
    print("Sentiment:", sentiment)
```

```python
# 3

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

nltk.download("stopwords", quiet=True)

reviews = [
    "The food was delicious and delivery was fast",
    "Amazing taste and excellent service",
    "I loved the food and the delivery",
    "The meal was fresh and very tasty",
    "Great food and friendly delivery person",
    "The food was cold and tasteless",
    "Very bad service and late delivery",
    "I hated the food, it was terrible",
    "Worst meal ever, completely disappointing",
    "The delivery was slow and the food was awful",
    "Good quality food and quick service",
    "The food was not good",
    "Excellent food and fast delivery",
    "The order was damaged and disgusting",
    "Really tasty food, I enjoyed it"
]

labels = [
    "Positive", "Positive", "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative", "Negative", "Negative",
    "Positive", "Negative", "Positive", "Negative", "Positive"
]

stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

```

```python

stop_words -= {"not", "no", "nor", "never"}

def preprocess(text):

    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

    words = [word for word in words if word not in stop_words]

    words = [stemmer.stem(word) for word in words]
    return " ".join(words)

X_train_text, X_test_text, y_train, y_test = train_test_split(
    reviews,
    labels,
    test_size=0.3,
    random_state=42,
    stratify=labels
)

X_train_clean = [preprocess(review) for review in X_train_text]
X_test_clean = [preprocess(review) for review in X_test_text]

vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(X_train_clean)
X_test = vectorizer.transform(X_test_clean)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)
print(f"Model Accuracy: {accuracy:.2f} ({accuracy * 100:.1f}%)")

for i in range(3):
    new_review = input(f"Enter review {i + 1}: ")
    clean_review = preprocess(new_review)
    review_vector = vectorizer.transform([clean_review])
    sentiment = model.predict(review_vector)[0]
    print("Predicted Sentiment:", sentiment)
```

4. Testing and debugging note

I found that the original code removed the word “not” during stopword removal, which could change the meaning of negative reviews. I fixed this by keeping negation words such as “not”, “no”, and “never”. I also moved TF-IDF fitting until after the train-test split to avoid data leakage. The corrected version gives a more reliable evaluation process, although the small dataset limits the model's accuracy and generalisation.
